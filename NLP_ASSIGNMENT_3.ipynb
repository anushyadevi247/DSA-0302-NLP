**TASK 2 – CFG Rules, Derivations and Parse Trees**

```python
from nltk import CFG
from nltk.parse import RecursiveDescentParser
grammar = CFG.fromstring("""
S -> NP VP
NP -> DET N
VP -> V NP
DET -> 'the'
N -> 'student' | 'timetable'
V -> 'checks'
""")
sentence = "the student checks the timetable".split()
left = [
    "S",
    "NP VP",
    "DET N VP",
    "the N VP",
    "the student VP",
    "the student V NP",
    "the student checks NP",
    "the student checks DET N",
    "the student checks the N",
    "the student checks the timetable"
]

right = [
    "S",
    "NP VP",
    "NP V NP",
    "NP V DET N",
    "NP V DET timetable",
    "NP V the timetable",
    "NP checks the timetable",
    "DET N checks the timetable",
    "the N checks the timetable",
    "the student checks the timetable"
]
print("LEFTMOST DERIVATION")
for step in left:
    print(step)
print("\nRIGHTMOST DERIVATION")
for step in right:
    print(step)
parser = RecursiveDescentParser(grammar)
print("\nPARSE TREE")
for tree in parser.parse(sentence):
    print(tree)
    tree.pretty_print()
```

**TASK 3 – Sentence-Level Constructions**

```python
def classify_sentence(sentence)
    words = sentence.lower().strip(" .?!").split()
    if words[0] in ["what", "where", "when", "who", "which"]:
        return "WH Question"
    elif words[0] in ["is", "are", "do", "does", "can", "will"]:
        return "Yes/No Question"
    elif words[0] in ["submit", "show", "check", "pay", "book"]:
        return "Imperative Sentence"
    elif "and" in words or "or" in words or "but" in words:
        return "Coordinated Sentence"
    else:
        return "Declarative Sentence"
sentences = [
    "The student pays the fee.",
    "Submit the assignment.",
    "Does the student check the timetable?",
    "Where is the library?",
    "The student pays the fee and hostel charges."
]
for sentence in sentences:
    print(sentence)
    print("Type:", classify_sentence(sentence))
    print()
```

**TASK 4 – Agreement**

```python
def check_agreement(sentence):
    words = sentence.lower().strip(" .?!").split()
    valid = [
        ("student", "submits"),
        ("student", "checks"),
        ("students", "submit"),
        ("students", "check"),
        ("he", "checks"),
        ("she", "checks"),
        ("they", "check")
    ]
    invalid = [
        ("students", "submits"),
        ("student", "submit"),
        ("he", "submit"),
        ("they", "submits")
    ]
    for subject, verb in valid:
        if subject in words and verb in words:
            return "ACCEPT"
    for subject, verb in invalid:
        if subject in words and verb in words:
            return "REJECT: Subject-Verb Disagreement"
    if words[0] == "him":
        return "REJECT: Incorrect Pronoun Case"

    return "CHECK: Pattern not covered"
tests = [
    "The student submits the form.",
    "The students submit the form.",
    "The students submits the form.",
    "The student submit the form.",
    "Him checks the result."
]
for sentence in tests:
    print(sentence, ":", check_agreement(sentence))
```

**TASK 5 – Subcategorization**

```python
frames = {
    "check": ["NP"],
    "submit": ["NP"],
    "pay": ["NP"],
    "arrive": [],
    "depend": ["PP"],
    "send": ["NP", "NP"],
    "place": ["NP", "PP"]
}
def check_frame(verb, complement):
    if verb not in frames:
        return "Unknown Verb"
    if frames[verb] == complement:
        return "Valid Subcategorization"
    return "Invalid Subcategorization"
tests = [
    ("submit", ["NP"]),
    ("arrive", []),
    ("depend", ["PP"]),
    ("send", ["NP", "NP"]),
    ("arrive", ["NP"])
]
for verb, complement in tests:
    print(verb, complement)
    print(check_frame(verb, complement))
    print()
```

**TASK 6 – Top-Down Parsing**

```python
from nltk import CFG
from nltk.parse import RecursiveDescentParser
grammar = CFG.fromstring("""
S -> NP VP
NP -> DET N
VP -> V NP
DET -> 'the'
N -> 'student' | 'timetable'
V -> 'checks'
""")
sentence = "the student checks the timetable".split()
parser = RecursiveDescentParser(grammar)
found = False
for tree in parser.parse(sentence):
    found = True
    print("Sentence Accepted")
    print(tree)
if not found:
    print("Sentence Rejected")
```

**TASK 7 – Earley Parsing**

```python
from nltk import CFG
from nltk.parse import EarleyChartParser
grammar = CFG.fromstring("""
S -> NP VP
NP -> DET N
VP -> V NP
DET -> 'the'
N -> 'student' | 'timetable'
V -> 'checks'
""")
sentence = "the student checks the timetable".split()
parser = EarleyChartParser(grammar)
trees = list(parser.parse(sentence))
print("Number of Parses:", len(trees))
for tree in trees:
    print(tree)
chart = parser.chart_parse(sentence)
print("\nEARLEY CHART STATES")
for i in range(len(sentence) + 1):
    states = list(chart.select(end=i))
    print("Chart", i, ":", len(states), "states")
```

**TASK 8 – Feature Structures and Unification**

```python
def unify(fs1, fs2):
    result = dict(fs1)
    for feature, value in fs2.items():
        if feature in result and result[feature] != value:
            return None
        result[feature] = value
    return result
student = {
    "NUM": "singular",
    "PERSON": "third",
    "CASE": "nominative"
}
verb = {
    "NUM": "singular",
    "PERSON": "third"
}
plural = {
    "NUM": "plural",
    "PERSON": "third"
}
print("Successful Unification:")
print(unify(student,verb))
print("\nFailed Unification:")
print(unify(student, plural))
```

**TASK 9 – Probabilistic Context-Free Grammar**

```python
from nltk import PCFG
from nltk.parse import ViterbiParser
grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> DET N [0.8] | PRON [0.2]
VP -> V NP [0.8] | V [0.2]
DET -> 'the' [0.7] | 'a' [0.3]
N -> 'student' [0.4] | 'form' [0.3] | 'timetable' [0.3]
PRON -> 'he' [0.5] | 'they' [0.5]
V -> 'checks' [0.5] | 'submits' [0.5]
""")
sentence = "the student checks the form".split()
parser = ViterbiParser(grammar)
trees = list(parser.parse(sentence))
if trees:
    best = trees[0]
    print("Best Parse Tree:")
    print(best)
    print("Probability:", best.prob())
else:
    print("No Parse Found")
```

**TASK 10 – Implement and Test the Parsers**

```python
import time
from nltk import CFG, PCFG
from nltk.parse import RecursiveDescentParser
from nltk.parse import EarleyChartParser
from nltk.parse import ViterbiParser
cfg = CFG.fromstring("""
S -> NP VP
NP -> DET N | PRON
VP -> V NP | V
DET -> 'the' | 'a'
N -> 'student' | 'students' | 'form' | 'timetable'
PRON -> 'he' | 'they'
V -> 'checks' | 'check' | 'submits' | 'submit'
""")
pcfg = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> DET N [0.8] | PRON [0.2]
VP -> V NP [0.8] | V [0.2]
DET -> 'the' [0.7] | 'a' [0.3]
N -> 'student' [0.4] | 'students' [0.2] | 'form' [0.2] | 'timetable' [0.2]
PRON -> 'he' [0.5] | 'they' [0.5]
V -> 'checks' [0.3] | 'check' [0.3] | 'submits' [0.2] | 'submit' [0.2]
""")
sentences = [
    "the student checks the timetable",
    "the students check the form",
    "he submits the form",
    "they check the timetable"
]
for sentence in sentences:
    words = sentence.split()
    print("\nInput:", sentence)
    start = time.time()
    td = list(RecursiveDescentParser(cfg).parse(words))
    td_time = time.time() - start
    start = time.time()
    earley = list(EarleyChartParser(cfg).parse(words))
    earley_time = time.time() - start
    start = time.time()
    vit = list(ViterbiParser(pcfg).parse(words))
    vit_time = time.time() - start
    print("Top-Down Parses:", len(td))
    print("Top-Down Time:", td_time)
    print("Earley Parses:", len(earley))
    print("Earley Time:", earley_time)
    print("Viterbi Probability:",
          vit[0].prob() if vit else "No Parse")
    print("Viterbi Time:", vit_time)
```

**TASK 11 – Compare and Recommend the Parsing Design**

```python
methods = {
    "Top-Down Parser": [
        "Starts from the start symbol S",
        "Uses grammar expansion",
        "May require backtracking",
        "Can have problems with left recursion"
    ],
    "Earley Parser": [
        "Uses a chart-based approach",
        "Uses Predictor, Scanner and Completer",
        "Handles left recursion",
        "Can store multiple parses"
    ],
    "PCFG / Viterbi Parser": [
        "Uses production probabilities",
        "Ranks alternative parse trees",
        "Selects the most probable parse",
        "Requires training data"
    ]
}
for method, details in methods.items():
    print("\n", method)
    for detail in details:
        print("-", detail)
print("\nRecommended Architecture:")
print("Feature-Based CFG + Earley Parser + PCFG Ranking")
```

**TASK 12 – Design Challenge**

```python
import re
def preprocess(sentence):
    sentence = sentence.lower()
    sentence = re.sub(r"[?.!,]", "", sentence)
    sentence = re.sub(r"\s+", " ", sentence)
    return sentence.strip()
def identify_topic(sentence):
    sentence = preprocess(sentence)
    topics = {
        "Exam": ["exam", "exams", "examination", "timetable", "result"],
        "Hostel": ["hostel", "room"],
        "Fee": ["fee", "fees", "payment", "receipt"],
        "Library": ["library", "book", "borrow", "return"],
        "Placement": ["placement", "job", "interview"]
    }
    found = []
    for topic, keywords in topics.items():
        for word in keywords:
            if word in sentence.split():
                found.append(topic)
                break
    if found:
        return found
    return ["Unknown Topic"]
messages = [
    "Can you tell me when the exams will be held?",
    "The hostel fee is due and the examination is postponed.",
    "Hi tell me exam date",
    "And the hostel fee?"
]
for message in messages:
    print("\nOriginal Message:", message)
    clean = preprocess(message)
    print("Preprocessed Message:", clean)
    print("Detected Topic:", identify_topic(clean))
    print("System: Extended grammar and parser can process the full query.")
```
