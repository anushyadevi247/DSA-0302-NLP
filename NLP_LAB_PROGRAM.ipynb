**1. Regular Expression Match and Search**

```python
import re

text = "Python is a programming language. My phone number is 9876543210."

# Search for a word
result = re.search("Python", text)

if result:
    print("Pattern Found:", result.group())
else:
    print("Pattern Not Found")

# Match a pattern at the beginning
result = re.match("Python", text)

if result:
    print("Match Found: YES")
else:
    print("Match Found: NO")

# Find all numbers
numbers = re.findall(r'\d+', text)
print("Numbers Found:", numbers)

# Replace a word
new_text = re.sub("Python", "Java", text)
print("After Replacement:", new_text)
```

**2. Finite State Automaton (Strings Ending with 'ab')**

```python
# Finite State Automaton to recognize strings ending with "ab"

def automaton(string):
    state = "q0"
    path = [state]

    for ch in string:
        if state == "q0":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q0"

        elif state == "q1":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q2"

        elif state == "q2":
            if ch == "a":
                state = "q1"
            elif ch == "b":
                state = "q0"

        path.append(state)

    print("State Path:", " -> ".join(path))

    if state == "q2":
        print("ACCEPTED")
    else:
        print("REJECTED")


string = input("Enter a string using a and b: ")

if all(ch in "ab" for ch in string):
    automaton(string)
else:
    print("Invalid input! Use only a and b.")
```

**3. Morphological Analysis using NLTK**

```python
import nltk
from nltk.stem import PorterStemmer, WordNetLemmatizer

nltk.download('wordnet')
nltk.download('omw-1.4')

words = ["playing", "played", "studies", "running", "cars", "better"]

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

print("Word\t\tStemming\tLemmatization")

for word in words:
    stem = stemmer.stem(word)
    lemma = lemmatizer.lemmatize(word)
    print(word, "\t\t", stem, "\t\t", lemma)
```

**4. Finite State Machine for Morphological Parsing**

```python
# Finite State Machine to generate plural nouns

def plural(noun):
    if noun.endswith(("s", "sh", "ch", "x", "z")):
        state = "q1"
        plural_word = noun + "es"

    elif noun.endswith("y") and noun[-2] not in "aeiou":
        state = "q2"
        plural_word = noun[:-1] + "ies"

    else:
        state = "q3"
        plural_word = noun + "s"

    print("Input Noun:", noun)
    print("State:", state)
    print("Plural Form:", plural_word)


noun = input("Enter a noun: ").lower()

if len(noun) > 1 and noun.isalpha():
    plural(noun)
else:
    print("Enter a valid noun")
```

**6. Bigram Model for Text Generation**

```python
import random
from collections import defaultdict, Counter

# Training text
text = "python is easy python is powerful python is useful"

# Split text into words
words = text.split()

# Create Bigram Model
bigram = defaultdict(Counter)

for i in range(len(words) - 1):
    bigram[words[i]][words[i + 1]] += 1

# Generate text
word = "python"
result = [word]

for i in range(5):
    next_words = list(bigram[word].keys())
    if not next_words:
        break

    word = random.choice(next_words)
    result.append(word)

print("Generated Text:")
print(" ".join(result))
```

**7. Part-of-Speech (POS) Tagging using NLTK**

```python
import nltk
from nltk.tokenize import word_tokenize

# Download required resources
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')

# Input text
text = input("Enter a sentence: ")

# Tokenize the text
words = word_tokenize(text)

# Perform POS tagging
pos_tags = nltk.pos_tag(words)

# Display output
print("\nPart of Speech Tagging:")
for word, tag in pos_tags:
    print(word, ":", tag)
```

**8. Stochastic Part-of-Speech Tagging using Python**

```python
import nltk
from nltk import word_tokenize
from nltk.probability import FreqDist, ConditionalFreqDist

# Training data
training_data = [
    [("The", "DT"), ("cat", "NN"), ("runs", "VB")],
    [("A", "DT"), ("dog", "NN"), ("barks", "VB")],
    [("The", "DT"), ("boy", "NN"), ("plays", "VB")],
    [("A", "DT"), ("girl", "NN"), ("sings", "VB")]
]

# Calculate word-tag probabilities
word_tags = {}

for sentence in training_data:
    for word, tag in sentence:
        word_tags.setdefault(word.lower(), FreqDist())
        word_tags[word.lower()][tag] += 1

# Stochastic POS tagging
def pos_tagging(words):
    result = []

    for word in words:
        if word.lower() in word_tags:
            tag = word_tags[word.lower()].max()
        else:
            tag = "NN"

        result.append((word, tag))

    return result

# Input
text = input("Enter a sentence: ")
words = word_tokenize(text)

# Output
print("\nPOS Tagging:")
for word, tag in pos_tagging(words):
    print(word, ":", tag)
```

**9. Rule-Based Part-of-Speech Tagging using Regular Expressions**

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk import RegexpTagger

# Input sentence
text = input("Enter a sentence: ")

# Tokenize the sentence
words = word_tokenize(text)

# Define regular expression rules
patterns = [
    (r'.*ing$', 'VBG'),       # Verb ending with ing
    (r'.*ed$', 'VBD'),        # Past tense verb
    (r'.*ly$', 'RB'),         # Adverb
    (r'.*ous$', 'JJ'),        # Adjective
    (r'.*', 'NN')             # Default noun
]

# Create rule-based tagger
tagger = RegexpTagger(patterns)

# Perform POS tagging
result = tagger.tag(words)

# Display output
print("\nPOS Tagging:")
for word, tag in result:
    print(word, ":", tag)
```

**10. Transformation-Based Tagging using Python**

```python
import nltk
from nltk.tokenize import word_tokenize

# Input sentence
text = input("Enter a sentence: ")

# Tokenize the sentence
words = word_tokenize(text)

# Initial tagging
tagged_words = []

for word in words:
    if word.lower() in ["is", "am", "are", "was", "were"]:
        tag = "VB"
    else:
        tag = "NN"

    tagged_words.append([word, tag])

# Transformation rules
for item in tagged_words:
    word = item[0]

    if word.endswith("ing"):
        item[1] = "VBG"

    elif word.endswith("ly"):
        item[1] = "RB"

    elif word[0].isupper():
        item[1] = "NNP"

# Display output
print("\nTransformation-Based POS Tagging:")

for word, tag in tagged_words:
    print(word, ":", tag)
```
