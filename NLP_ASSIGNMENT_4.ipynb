**Task 2 – The Meaning Structure of Language**

```python
knowledge_base = {
    "Ravi": 12,
    "Meena": 8
}
def check_statement(name, required_years):
    if name not in knowledge_base:
        return "UNKNOWN"
    if knowledge_base[name] >= required_years:
        return "TRUE"
    else:
        return "FALSE"
print(check_statement("Ravi", 10))
print(check_statement("Meena", 10))
print(check_statement("Arun", 10))
```

**Task 3 – First Order Predicate Calculus**

```python
employees = {
    "Ravi": {"service": 12, "dependent": True},
    "Meena": {"service": 8, "dependent": True},
    "Arun": {"service": 15, "dependent": False}
}
def employee(name):
    return name in employees
def qualifying_service(name):
    return employees[name]["service"] >= 10
def dependent(name):
    return employees[name]["dependent"]
def eligible(name):
    return (
        employee(name)
        and qualifying_service(name)
        and dependent(name)
    )
for name in employees:
    print(name, eligible(name))
```

**Task 7 – Word Sense Disambiguation**

```python
import re
senses = {
    "service": {
        "employment": "government employee work years retirement service",
        "support": "customer help assistance service"
    }
}
def words(text):
    return set(re.findall(r"[a-z]+", text.lower()))
def lesk(context, word):
    context_words = words(context)
    best_sense = None
    best_score = -1
    for sense, gloss in senses[word].items():
        score = len(context_words & words(gloss))
        if score > best_score:
            best_score = score
            best_sense = sense
    return best_sense, best_score
sentence = "The employee completed twenty years of government service."
sense, score = lesk(sentence, "service")
print("Selected Sense:", sense)
print("Overlap Score:", score)
```

**Task 8 – Information Retrieval over the Circulars**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
documents = [
    "Pension eligibility requires qualifying service of ten years.",
    "Family pension is provided to eligible dependents.",
    "Pension commutation calculation depends on age and pension amount.",
    "Government circular explains pension revision rules."
]
vectorizer = TfidfVectorizer(stop_words="english")
document_matrix = vectorizer.fit_transform(documents)
query = ["qualifying service for pension eligibility"]
query_vector = vectorizer.transform(query)
scores = cosine_similarity(query_vector, document_matrix)
ranked = scores[0].argsort()[::-1]
for index in ranked:
    print("Circular:", documents[index])
    print("Similarity:", round(scores[0][index], 3))
```

**Task 9 – Annotation, Gold Standard and Evaluation Design**

```python
from sklearn.metrics import cohen_kappa_score
annotator1 = [
    "employment",
    "financial",
    "family",
    "employment",
    "financial"
]
annotator2 = [
    "employment",
    "financial",
    "family",
    "support",
    "financial"
]
agreement = sum(
    a == b for a, b in zip(annotator1, annotator2)
) / len(annotator1)
kappa = cohen_kappa_score(annotator1, annotator2)
print("Raw Agreement:", round(agreement, 2))
print("Cohen Kappa:", round(kappa, 2))
```

**Task 10 – Implement and Test the Pipeline**

```python
import re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
documents = [
    "Pension eligibility requires ten years of qualifying service.",
    "Family pension is provided to eligible dependents.",
    "Pension commutation depends on age and pension amount."
]
def preprocess(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    return text
clean_documents = [preprocess(d) for d in documents]
def check_eligibility(years):
    if years >= 10:
        return "Eligible"
    return "Not Eligible"
vectorizer = TfidfVectorizer(stop_words="english")
matrix = vectorizer.fit_transform(clean_documents)
def search(query):
    query = preprocess(query)
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, matrix)[0]
    for i in scores.argsort()[::-1]:
        print(documents[i], round(scores[i], 3))
print("Eligibility:", check_eligibility(12))
print("\nRetrieved Circulars:")
search("qualifying service pension eligibility")
```

**Task 12 – Re-evaluate and Adapt the Design**

```python
from datetime import date
def pension_eligibility(
    service_periods,
    retirement_date,
    minimum_years
):
    qualifying_days = 0
    for start, end in service_periods:
        start_date = date.fromisoformat(start)
        end_date = date.fromisoformat(end)
        if end_date < start_date:
            raise ValueError("Invalid service period")
        qualifying_days += (end_date - start_date).days
    qualifying_years = qualifying_days / 365.25
    if qualifying_years >= minimum_years:
        return "Eligible"
    else:
        return "Not Eligible"
periods = [
    ("2000-01-01", "2010-01-01"),
    ("2012-01-01", "2015-01-01")
]
result = pension_eligibility(
    periods,
    "2019-01-01",
    10
)
print("Pension Status:", result)
```
